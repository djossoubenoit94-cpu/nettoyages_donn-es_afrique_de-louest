<h2>01 | CONTEXTE, PROBLÉMATIQUE ET OBJECTIFS</h2>

</div>
<h3 id="s01-1">1.1. Contexte et justification de l'étude<span class="up"><a href="#section-01">↑ section 01</a><a href="#sommaire">☰ sommaire</a></span></h3>
<div class="card">
<p>Les températures historiques servent à suivre l'évolution du climat, mais les fichiers bruts sont rarement utilisables tels quels : dates enregistrées comme du texte, mesures absentes, noms de pays écrits de plusieurs façons.</p>
<p>Nous travaillons sur le fichier <code>GlobalLandTemperaturesByCountry.csv</code> (Berkeley Earth, disponible sur Kaggle). Comme l'énoncé le demande, nous n'utilisons que ce fichier et nous gardons cinq pays d'Afrique de l'Ouest.</p>
</div>
<h3 id="s01-2">1.2. Problématique<span class="up"><a href="#section-01">↑ section 01</a><a href="#sommaire">☰ sommaire</a></span></h3>
<div class="card">
<p>Nettoyer trop vite peut fausser une série historique : si l'on supprime toutes les lignes incomplètes ou si l'on remplit les trous avec une moyenne, on risque d'inventer des températures ou d'effacer des périodes entières. Il faut donc faire la différence entre une vraie erreur, une simple absence de mesure et une mesure peu précise, et expliquer chaque choix.</p>
</div>
<div class="question"><strong>Question centrale :</strong> comment repérer et traiter les problèmes de qualité de ces températures sans perdre ni inventer d'information ?</div>
<h3 id="s01-3">1.3. Objectif général<span class="up"><a href="#section-01">↑ section 01</a><a href="#sommaire">☰ sommaire</a></span></h3>
<div class="objective">Obtenir, pour cinq pays d'Afrique de l'Ouest, un fichier de températures propre, cohérent et documenté, prêt pour une analyse exploratoire.</div>
<h3 id="s01-4">1.4. Objectifs spécifiques<span class="up"><a href="#section-01">↑ section 01</a><a href="#sommaire">☰ sommaire</a></span></h3>
<div class="card"><ul>
<li>décrire le fichier d'origine (dimensions, colonnes, types) ;</li>
<li>retrouver nos cinq pays sous leurs noms exacts ;</li>
<li>convertir les dates et vérifier la chronologie ;</li>
<li>mesurer les valeurs manquantes, les doublons, les incertitudes et les valeurs douteuses ;</li>
<li>justifier chaque correction et comparer l'état avant et après ;</li>
<li>livrer ce notebook commenté et un CSV nettoyé.</li>
</ul></div>
<h3 id="s01-5">1.5. Démarche méthodologique<span class="up"><a href="#section-01">↑ section 01</a><a href="#sommaire">☰ sommaire</a></span></h3>
<div class="card">
<p>Nous commençons par charger le fichier en gardant l'original intact, puis nous isolons les cinq pays (<a href="#section-04">section 04</a>) et nous faisons un premier diagnostic (<a href="#section-05">section 05</a>). Vient ensuite le nettoyage proprement dit : dates, noms, doublons, manquants, incertitudes et valeurs aberrantes (<a href="#section-06">sections 06</a> à <a href="#section-10">10</a>). Nous recalculons enfin les mêmes indicateurs sur le résultat (<a href="#section-11">section 11</a>), nous vérifions le tout par des graphiques, puis nous exportons le CSV (<a href="#section-13">section 13</a>). Tout est fait en Python avec pandas, dans Jupyter.</p>
</div>
<h3 id="s01-6">1.6. Résultats attendus<span class="up"><a href="#section-01">↑ section 01</a><a href="#sommaire">☰ sommaire</a></span></h3>
<div class="card"><ul>
<li>un état des lieux chiffré avant toute correction (<a href="#section-05">section 05</a>) ;</li>
<li>un tableau des corrections avec leurs justifications (<a href="#s14-1">14.1</a>) ;</li>
<li>un état final comparable à l'état initial (<a href="#s11-2">11.2</a>) ;</li>
<li>le CSV nettoyé, limité aux cinq pays (<a href="#section-13">section 13</a>).</li>
</ul></div>
<div class="note"><strong>Notre règle :</strong> une valeur manquante n'est pas forcément une erreur, et une incertitude élevée ne rend pas une mesure fausse. Nous ne supprimons et nous n'imputons rien sans l'avoir justifié.</div>
</div>

<h2>02 | PRÉSENTATION DU JEU DE DONNÉES</h2>

</div>
<h3 id="s02-1">2.1. Source et nature des données<span class="up"><a href="#section-02">↑ section 02</a><a href="#sommaire">☰ sommaire</a></span></h3>
<div class="card">
<p>Les données viennent de Berkeley Earth, publiées sur Kaggle : <a href="https://www.kaggle.com/datasets/berkeleyearth/climate-change-earth-surface-temperature-data" target="_blank">Climate Change: Earth Surface Temperature Data</a>. Seul le fichier <code>GlobalLandTemperaturesByCountry.csv</code> est utilisé. Il contient des températures terrestres moyennes mensuelles, pays par pays.</p>
</div>
<h3 id="s02-2">2.2. Structure et signification des variables<span class="up"><a href="#section-02">↑ section 02</a><a href="#sommaire">☰ sommaire</a></span></h3>
<div class="card">
<table>
<thead><tr><th>Variable</th><th>Signification</th><th>Type visé</th></tr></thead>
<tbody>
<tr><td><code>dt</code></td><td>Date de l'observation</td><td><code>datetime64</code></td></tr>
<tr><td><code>AverageTemperature</code></td><td>Température moyenne du mois, en °C</td><td><code>float</code></td></tr>
<tr><td><code>AverageTemperatureUncertainty</code></td><td>Incertitude de cette estimation, en °C</td><td><code>float</code></td></tr>
<tr><td><code>Country</code></td><td>Nom du pays</td><td><code>string</code></td></tr>
</tbody></table>
<p>Il s'agit donc de séries mensuelles, une par pays.</p>
</div>
<h3 id="s02-3">2.3. Périmètre géographique et temporel<span class="up"><a href="#section-02">↑ section 02</a><a href="#sommaire">☰ sommaire</a></span></h3>
<div class="card">
<p>Nous gardons le Bénin, le Burkina Faso, la Côte d'Ivoire, le Ghana et le Sénégal. Le fichier complet compte 577 462 lignes pour 243 entités géographiques ; après sélection, il reste 9 885 lignes, soit 1 977 mois par pays, de janvier 1849 à septembre 2013 (vérifié à la <a href="#section-06">section 06</a>).</p>
</div>
<h3 id="s02-4">2.4. Problèmes de qualité à examiner<span class="up"><a href="#section-02">↑ section 02</a><a href="#sommaire">☰ sommaire</a></span></h3>
<div class="card">
<table>
<thead><tr><th>Problème signalé dans l'énoncé</th><th>Ce que nous vérifions</th><th>Section</th></tr></thead>
<tbody>
<tr><td>Dates à convertir</td><td>Format de <code>dt</code> et dates invalides</td><td><a href="#section-06">06</a></td></tr>
<tr><td>Mesures anciennes manquantes</td><td>Manquants par pays et par période</td><td><a href="#section-05">05</a> et <a href="#section-08">08</a></td></tr>
<tr><td>Incertitudes élevées au début</td><td>Distribution et évolution dans le temps</td><td><a href="#section-09">09</a></td></tr>
<tr><td>Pays présents sous deux noms</td><td>Variantes de noms à harmoniser</td><td><a href="#section-07">07</a></td></tr>
</tbody></table>
<p>Nous regardons aussi les doublons, les conflits pays-date, les valeurs impossibles et d'éventuels mois absents du calendrier.</p>
</div>
<h3 id="s02-5">2.5. Principes de traitement retenus<span class="up"><a href="#section-02">↑ section 02</a><a href="#sommaire">☰ sommaire</a></span></h3>
<div class="card">
<p>Nous gardons le fichier d'origine intact et nous notons chaque transformation. Nous ne considérons pas d'office une valeur manquante ou une incertitude élevée comme une erreur. Les mêmes indicateurs sont calculés avant et après le nettoyage pour mesurer ce qui a changé, et le tout doit pouvoir être relancé d'un bout à l'autre.</p>
</div>
</div>

```python
# Références du jeu de données
dataset = "berkeleyearth/climate-change-earth-surface-temperature-data"   # Identifiant Kaggle
nom_fichier = "GlobalLandTemperaturesByCountry.csv"                       # Seul fichier autorisé par la consigne

# Utiliser le fichier local s'il existe, sinon le télécharger depuis Kaggle
chemin_local = Path(nom_fichier)
chemin_csv = chemin_local if chemin_local.exists() else kagglehub.dataset_download(dataset, path=nom_fichier)

df_brut = pd.read_csv(chemin_csv)   # Données originales : on n'y touchera plus
df = df_brut.copy(deep=True)        # Copie indépendante pour préserver l'original
```

<h3 id="s04-2">4.2. Vérification des cinq pays de l'étude<span class="up"><a href="#section-04">↑ section 04</a><a href="#sommaire">☰ sommaire</a></span></h3>
<div class="card"><p>Les noms sont cherchés tels qu'on les écrit habituellement. Le fichier utilise l'orthographe anglaise et une casse particulière pour la Côte d'Ivoire (<code>Côte D'Ivoire</code>, avec un « D » majuscule) : le test peut donc afficher <code>False</code> pour certaines variantes, ce qui est normal tant qu'une variante est trouvée.</p></div>

Le code s'arrête avec une erreur si l'un des cinq pays manque, ce qui garantit la consigne « exactement cinq pays ».</p></div>
</div>

```python
# Dictionnaire : variante trouvée dans le fichier -> libellé harmonisé (français)
correspondance_pays = {
    "Benin": "Bénin",
    "Burkina Faso": "Burkina Faso",
    "Côte D'Ivoire": "Côte d'Ivoire",
    "Côte d'Ivoire": "Côte d'Ivoire",
    "Cote D'Ivoire": "Côte d'Ivoire",
    "Ivory Coast": "Côte d'Ivoire",
    "Ghana": "Ghana",
    "Senegal": "Sénégal",
}

# Noms réellement présents dans le fichier parmi les variantes connues
pays_disponibles = set(df["Country"].dropna().unique())
noms_retenus = pays_disponibles.intersection(correspondance_pays)

# Vérifier que les cinq pays sont représentés
pays_identifies = {correspondance_pays[p] for p in noms_retenus}
if len(pays_identifies) != 5:
    raise ValueError(f"Pays manquants : {set(correspondance_pays.values()) - pays_identifies}")

# Extraire uniquement les observations des cinq pays : df_etude = état INITIAL de l'étude
df_etude = df[df["Country"].isin(noms_retenus)].copy()

print("Nombre de pays sélectionnés :", len(pays_identifies))
print("Nombre total d'observations :", len(df_etude))
print("Dimensions du jeu de données :", df_etude.shape)
print("Libellés présents dans le fichier :", sorted(noms_retenus))
```

<h3 id="s05-2">5.2. Valeurs manquantes par pays<span class="up"><a href="#section-05">↑ section 05</a><a href="#sommaire">☰ sommaire</a></span></h3>
<div class="card"><p>Les pays ne sont pas touchés de la même façon : on compare les taux de manquants pour mesurer l'ampleur du problème, sans rien corriger.</p></div>
</div>

```python
# Copie de diagnostic : df_etude reste strictement inchangé
df_diagnostic = df_etude.copy()

# Bilan des manquants par pays
# Regrouper les observations par pays et compter les valeurs manquantes.
bilan_pays = df_diagnostic.groupby("Country").agg(
    nombre_observations=("dt", "size"),
    temperatures_manquantes=("AverageTemperature", lambda s: s.isna().sum()),
    incertitudes_manquantes=("AverageTemperatureUncertainty", lambda s: s.isna().sum())
).reset_index()

# Calculer les pourcentages pour comparer équitablement les pays.
bilan_pays["taux_temperatures_manquantes_pct"] = (
    100 * bilan_pays["temperatures_manquantes"] / bilan_pays["nombre_observations"]
).round(2)
bilan_pays["taux_incertitudes_manquantes_pct"] = (
    100 * bilan_pays["incertitudes_manquantes"] / bilan_pays["nombre_observations"]
).round(2)

# Afficher le bilan initial par pays, sans corriger les données.
display(bilan_pays.sort_values("taux_temperatures_manquantes_pct", ascending=False))
```

<h3 id="s05-3">5.3. Valeurs manquantes par décennie<span class="up"><a href="#section-05">↑ section 05</a><a href="#sommaire">☰ sommaire</a></span></h3>
<div class="card"><p>Les mesures anciennes manquent-elles davantage ? On regroupe par décennie (1987 devient 1980). Les dates sont converties uniquement dans la copie de diagnostic.</p></div>
</div>

```python
# Convertir les dates uniquement dans la copie destinée au diagnostic.
df_diagnostic["date_analyse"] = pd.to_datetime(df_diagnostic["dt"], errors="coerce")

# Extraire les années puis les décennies : 1987 devient 1980.
df_diagnostic["annee"] = df_diagnostic["date_analyse"].dt.year
df_diagnostic["decennie"] = (df_diagnostic["annee"] // 10 * 10).astype("Int64")

# Regrouper par pays et décennie, y compris les éventuelles dates invalides.
bilan_historique = df_diagnostic.groupby(
    ["Country", "decennie"], dropna=False
).agg(
    nombre_observations=("dt", "size"),
    temperatures_manquantes=("AverageTemperature", lambda s: s.isna().sum()),
    incertitudes_manquantes=("AverageTemperatureUncertainty", lambda s: s.isna().sum())
).reset_index()

# Calculer les taux de valeurs manquantes par décennie.
bilan_historique["taux_temperatures_manquantes_pct"] = (
    100 * bilan_historique["temperatures_manquantes"] / bilan_historique["nombre_observations"]
).round(2)
bilan_historique["taux_incertitudes_manquantes_pct"] = (
    100 * bilan_historique["incertitudes_manquantes"] / bilan_historique["nombre_observations"]
).round(2)

# Afficher le tableau, classé dans l'ordre chronologique pour chaque pays.
display(bilan_historique.sort_values(["Country", "decennie"]))
```

<h3 id="s05-5">5.5. Premier repérage des valeurs douteuses<span class="up"><a href="#section-05">↑ section 05</a><a href="#sommaire">☰ sommaire</a></span></h3>
<div class="card"><p>L'énoncé demande aussi de repérer les valeurs aberrantes <em>avant</em> toute correction. Nous comptons ici, sur le jeu initial, quatre types de valeurs douteuses, avec des règles simples que nous réutiliserons aux sections <a href="#section-09">09</a> et <a href="#section-10">10</a> : une température hors de la plage 10 à 45 °C, une incertitude nulle ou négative, une température atypique pour son pays et son mois (règle de l'IQR), et une incertitude élevée (clôture de Tukey). On ne modifie rien, on compte seulement.</p></div>
</div>

```python
T_MIN, T_MAX = 10, 45     # plage de plausibilité d'une moyenne mensuelle en Afrique de l'Ouest (°C)

def compter_anomalies(d):
    """Compte les valeurs douteuses d'un jeu de données, sans rien modifier."""
    d = d[["dt", "Country", "AverageTemperature", "AverageTemperatureUncertainty"]].copy()
    d["mois"] = pd.to_datetime(d["dt"], errors="coerce").dt.month
    t = d["AverageTemperature"]
    u = d["AverageTemperatureUncertainty"]

    # quartiles de chaque groupe pays x mois (les saisons ne se comparent pas entre elles)
    g = d.groupby(["Country", "mois"])["AverageTemperature"]
    q1 = g.transform(lambda s: s.quantile(0.25))
    q3 = g.transform(lambda s: s.quantile(0.75))
    iqr = q3 - q1

    # clôture de Tukey pour l'incertitude, calculée sur tout le jeu
    q1u, q3u = u.quantile(0.25), u.quantile(0.75)

    return {
        "Températures hors plage (10 à 45 °C)": int(((t < T_MIN) | (t > T_MAX)).sum()),
        "Incertitudes nulles ou négatives": int((u <= 0).sum()),
        "Températures atypiques (IQR par pays et par mois)": int(((t < q1 - 1.5 * iqr) | (t > q3 + 1.5 * iqr)).sum()),
        "Incertitudes élevées (clôture de Tukey)": int((u > q3u + 1.5 * (q3u - q1u)).sum()),
    }

anomalies_initiales = compter_anomalies(df_etude)
display(pd.Series(anomalies_initiales, name="Jeu initial").to_frame())
```

<h2>06 | CONVERSION ET VALIDATION DES DATES</h2>

</div><h3 id="s06-1">6.1. Conversion de dt en datetime<span class="up"><a href="#section-06">↑ section 06</a><a href="#sommaire">☰ sommaire</a></span></h3><div class="card"><p>La colonne <code>dt</code> est lue comme du texte. Pour une série temporelle, on la convertit en vraie date (<code>datetime64</code>) : cela permet de trier, d'extraire l'année ou le mois et de calculer des durées. Avec <code>errors="coerce"</code>, toute date illisible devient <code>NaT</code>, ce qui permet de la compter au lieu de bloquer le code.</p></div>
</div>

```python
# Copie de travail du nettoyage : df_etude (état initial) reste intact pour la comparaison finale
df_net = df_etude.copy()

print("Type avant conversion :", df_net["dt"].dtype)

# Conversion en date ; format explicite AAAA-MM-JJ, les dates illisibles deviennent NaT
df_net["dt"] = pd.to_datetime(df_net["dt"], format="%Y-%m-%d", errors="coerce")

print("Type après conversion :", df_net["dt"].dtype)
print("Dates invalides (NaT) :", df_net["dt"].isna().sum())
print("Première date :", df_net["dt"].min().date())
print("Dernière date :", df_net["dt"].max().date())
```

<h3 id="s06-2">6.2. Validation des dates<span class="up"><a href="#section-06">↑ section 06</a><a href="#sommaire">☰ sommaire</a></span></h3>
<div class="card"><p>Trois contrôles : (1) toutes les dates tombent le 1er du mois (données mensuelles) ; (2) pas de doublon pays-date ; (3) continuité : le nombre de mois observés doit égaler le nombre de mois entre la première et la dernière date.</p></div>
</div>

```python
# 1) Données mensuelles : toutes les dates doivent tomber le 1er du mois
tous_premier = (df_net["dt"].dt.day == 1).all()
print("Toutes les dates au 1er du mois :", tous_premier)

# 2) Conflits : un même pays ne doit avoir qu'une seule ligne par date
n_doublons_pays_date = df_net.duplicated(subset=["Country", "dt"]).sum()
print("Doublons pays-date :", n_doublons_pays_date)

# 3) Continuité : mois attendus (début -> fin) comparés aux mois réellement présents
controle = df_net.groupby("Country")["dt"].agg(debut="min", fin="max", observes="size")
controle["attendus"] = ((controle["fin"].dt.year - controle["debut"].dt.year) * 12
                        + controle["fin"].dt.month - controle["debut"].dt.month + 1)
controle["mois_absents"] = controle["attendus"] - controle["observes"]

# 4) Chronologie : les lignes de chaque pays sont-elles déjà dans l'ordre des dates ?
controle["trie"] = df_net.groupby("Country")["dt"].apply(lambda s: s.is_monotonic_increasing)
display(controle)

# Verdict automatique
if tous_premier and n_doublons_pays_date == 0 and (controle["mois_absents"] == 0).all():
    print("OK : dates valides, séries mensuelles complètes (aucun mois absent du calendrier).")
else:
    print("ATTENTION : anomalie de dates à examiner avant de poursuivre.")
```

<h2>07 | HARMONISATION DES NOMS ET DOUBLONS</h2>

</div><h3 id="s07-1">7.1. Harmonisation des libellés de pays<span class="up"><a href="#section-07">↑ section 07</a><a href="#sommaire">☰ sommaire</a></span></h3><div class="card"><p>L'énoncé signale des pays « présents sous deux noms ». Pour nos cinq pays, le fichier n'utilise qu'un libellé par pays (5 libellés distincts, voir <a href="#s04-3">4.3</a>) : il n'y a donc pas de pays dupliqué à fusionner. En revanche, les libellés sont en anglais et la Côte d'Ivoire a une casse particulière (<code>Côte D'Ivoire</code>). On les harmonise en français à l'aide du dictionnaire de la section 4.3, pour obtenir des noms propres et stables dans le fichier final.</p></div>
</div>

```python
# Tableau des libellés AVANT harmonisation, avec le libellé cible
avant = (df_net["Country"].value_counts()
         .rename_axis("Libellé d'origine").reset_index(name="observations"))
avant["Libellé harmonisé"] = avant["Libellé d'origine"].str.strip().map(correspondance_pays)
avant["modifié"] = avant["Libellé d'origine"] != avant["Libellé harmonisé"]
display(avant)

# Espaces superflus éventuels autour des noms
print("Libellés avec espaces superflus :", (df_net["Country"] != df_net["Country"].str.strip()).sum())

# Application : suppression des espaces puis remplacement par le libellé harmonisé
# (fillna garde le libellé d'origine si jamais une variante n'était pas dans le dictionnaire)
df_net["Country"] = df_net["Country"].str.strip().map(correspondance_pays).fillna(df_net["Country"])

print("\nLibellés distincts après harmonisation :", df_net["Country"].nunique())
display(df_net["Country"].value_counts().rename("observations").to_frame())
```

<h3 id="s07-2">7.2. Recherche des doublons<span class="up"><a href="#section-07">↑ section 07</a><a href="#sommaire">☰ sommaire</a></span></h3>
<div class="card"><p>Après harmonisation, on recherche : (a) les doublons exacts (toutes colonnes identiques) ; (b) les doublons pays-date ; (c) parmi ceux-ci, les conflits (même pays et même date mais valeurs différentes), qui demanderaient un arbitrage. Un doublon exact peut être supprimé sans perte d'information ; un conflit ne se tranche pas automatiquement.</p></div>
</div>

```python
n_exacts = df_net.duplicated().sum()                                        # lignes strictement identiques
n_pays_date = df_net.duplicated(subset=["Country", "dt"]).sum()             # même pays et même date
n_conflits = n_pays_date - n_exacts                                         # même pays-date mais valeurs différentes

print("Doublons exacts :", n_exacts)
print("Doublons pays-date :", n_pays_date)
print("Conflits pays-date (valeurs différentes) :", n_conflits)

# Suppression des seuls doublons exacts (aucune perte d'information)
if n_exacts > 0:
    df_net = df_net.drop_duplicates().reset_index(drop=True)
    print(f"{n_exacts} doublon(s) exact(s) supprimé(s).")

if n_pays_date == 0:
    print("OK : aucun doublon, chaque pays a au plus une ligne par mois.")
elif n_conflits > 0:
    print("ATTENTION : conflits à arbitrer à la main :")
    display(df_net[df_net.duplicated(subset=["Country", "dt"], keep=False)].sort_values(["Country", "dt"]))
print("Lignes conservées :", len(df_net))
```

<h2>08 | TRAITEMENT DES VALEURS MANQUANTES</h2>

</div><h3 id="s08-1">8.1. Nature des valeurs manquantes<span class="up"><a href="#section-08">↑ section 08</a><a href="#sommaire">☰ sommaire</a></span></h3><div class="card"><p>Avant de décider quoi que ce soit, on cherche pourquoi les valeurs manquent : (1) les deux colonnes manquent-elles ensemble ? (2) les manquants sont-ils au <em>début</em> de la série (période sans mesure) ou en <em>trous</em> isolés ensuite ?</p></div>
</div>

```python
manq_t = df_net["AverageTemperature"].isna()               # température manquante
manq_u = df_net["AverageTemperatureUncertainty"].isna()    # incertitude manquante

print("Températures manquantes :", manq_t.sum())
print("Incertitudes manquantes :", manq_u.sum())
print("Manquantes simultanément :", (manq_t & manq_u).sum())
print("Manquant dans une seule des deux colonnes :", (manq_t ^ manq_u).sum())

# Profil des manquants de chaque pays : quand surviennent-ils ?
derniere_date = df_net["dt"].max()                           # dernier mois du fichier

def profil_manquants(g):
    manquants = g[g["AverageTemperature"].isna()]
    avant_fin = manquants[manquants["dt"] < derniere_date]   # on met à part le tout dernier mois
    return pd.Series({
        "première_mesure": g.loc[g["AverageTemperature"].notna(), "dt"].min().date(),
        "mois_manquants": len(manquants),
        "dont_avant_1900": int((manquants["dt"].dt.year < 1900).sum()),
        "dont_depuis_1900": int((manquants["dt"].dt.year >= 1900).sum()),
        "dernier_trou_avant_le_dernier_mois": avant_fin["dt"].max().date() if len(avant_fin) else None,
        "dernier_mois_sans_mesure": bool((manquants["dt"] == derniere_date).any()),
    })

profil = df_net.groupby("Country")[["dt", "AverageTemperature"]].apply(profil_manquants)
display(profil)
```

<h3 id="s08-2">8.2. Stratégies possibles et décision<span class="up"><a href="#section-08">↑ section 08</a><a href="#sommaire">☰ sommaire</a></span></h3>
<div class="card"><table>
<thead><tr><th>Option</th><th>Avantage</th><th>Inconvénient</th><th>Décision</th></tr></thead>
<tbody>
<tr><td>Supprimer les lignes</td><td>Simple ; aucune valeur inventée</td><td>Crée des trous dans le calendrier, fausse les moyennes annuelles et perd 775 lignes (7,84 %), dont plus de 99 % au XIXᵉ siècle</td><td>Non, pas dans le fichier nettoyé</td></tr>
<tr><td>Imputer par la moyenne ou la médiane</td><td>Aucun <code>NaN</code></td><td>Ignore la saisonnalité et la tendance : un mois de 1860 recevrait une valeur de 2000</td><td>Non</td></tr>
<tr><td>Interpoler</td><td>Respecte la continuité locale</td><td>Invente des mesures ; inadapté aux longues périodes vides du XIXᵉ siècle</td><td>Non</td></tr>
<tr><td>Conserver le <code>NaN</code> et le signaler</td><td>Aucune information inventée ; le calendrier reste complet ; l'utilisateur choisit son traitement</td><td>Les analyses doivent gérer les <code>NaN</code> (pandas le fait par défaut)</td><td>Retenu</td></tr>
</tbody></table>
<p>Les manquants sont des absences de mesure, concentrées dans le passé : les remplacer reviendrait à fabriquer des températures historiques. Nous conservons donc les lignes, nous laissons <code>NaN</code> et nous ajoutons un indicateur explicite.</p></div>
</div>

<h3 id="s08-4">8.4. Quand les mesures deviennent-elles fiables ?<span class="up"><a href="#section-08">↑ section 08</a><a href="#sommaire">☰ sommaire</a></span></h3>
<div class="card"><p>Pour guider un futur utilisateur, on compare les taux de manquants <em>avant 1900</em> et <em>depuis 1900</em>. Le seuil de 1900 est un repère indicatif ; la colonne <code>dernier_trou_avant_le_dernier_mois</code> de la section 8.1 donne la date exacte, pays par pays.</p></div>
</div>

```python
# Période : repère indicatif avant / depuis 1900
periode = np.where(df_net["dt"].dt.year < 1900, "avant 1900", "depuis 1900")

taux = (df_net.assign(periode=periode)
        .pivot_table(index="Country", columns="periode", values="temperature_manquante", aggfunc="mean") * 100)
display(taux.round(2).rename_axis(columns="Taux de températures manquantes (%)"))

# Que reste-t-il comme mois sans mesure depuis 1900 ?
depuis_1900 = df_net[(df_net["dt"].dt.year >= 1900) & df_net["temperature_manquante"]]
display(depuis_1900.groupby("Country")["dt"].agg(mois_sans_mesure="count", premier="min", dernier="max"))
```

<h3 id="s09-3">9.3. Marquage des incertitudes élevées<span class="up"><a href="#section-09">↑ section 09</a><a href="#sommaire">☰ sommaire</a></span></h3>
<div class="card"><p>Une incertitude élevée ne rend pas la mesure fausse : elle indique une moindre précision. On marque les observations concernées sans les supprimer, avec une règle objective : un mois est « à forte incertitude » si son incertitude dépasse <code>Q3 + 1,5 × IQR</code> (règle de Tukey, déjà utilisée pour les détecteurs IQR du module 13). L'utilisateur pourra ensuite choisir de les garder ou de les écarter selon son analyse.</p></div>
</div>

```python
inc = df_net["AverageTemperatureUncertainty"]
q1, q3 = inc.quantile(0.25), inc.quantile(0.75)
seuil_inc = q3 + 1.5 * (q3 - q1)                       # clôture haute de Tukey

# True si l'incertitude dépasse le seuil (les NaN donnent False)
df_net["incertitude_elevee"] = inc > seuil_inc

print(f"Seuil d'incertitude élevée : {seuil_inc:.3f} °C")
print("Mois à forte incertitude :", df_net["incertitude_elevee"].sum(),
      f"({df_net['incertitude_elevee'].mean() * 100:.2f} %)")

# Où se concentrent-ils ? Part de mois à forte incertitude par décennie
part_dec = (df_net.groupby("decennie")["incertitude_elevee"].mean() * 100).round(1)
display(part_dec.rename("% de mois à forte incertitude").to_frame())
```

<h2>10 | RECHERCHE DES VALEURS ABERRANTES</h2>

</div><h3 id="s10-1">10.1. Contrôle de plausibilité physique<span class="up"><a href="#section-10">↑ section 10</a><a href="#sommaire">☰ sommaire</a></span></h3><div class="card"><p>Premier niveau de contrôle : des valeurs physiquement impossibles. Pour une température moyenne mensuelle en Afrique de l'Ouest, nous retenons une plage très large de plausibilité de 10 °C à 45 °C ; une incertitude doit être strictement positive.</p></div>
</div>

```python
# Plage de plausibilité T_MIN / T_MAX définie en 5.5
hors_plage = df_net[(df_net["AverageTemperature"] < T_MIN) | (df_net["AverageTemperature"] > T_MAX)]
incert_invalides = (df_net["AverageTemperatureUncertainty"] <= 0).sum()

print(f"Températures hors de [{T_MIN} ; {T_MAX}] °C :", len(hors_plage))
print("Incertitudes nulles ou négatives :", incert_invalides)
print(f"Température minimale observée : {df_net['AverageTemperature'].min():.2f} °C")
print(f"Température maximale observée : {df_net['AverageTemperature'].max():.2f} °C")

if len(hors_plage) == 0 and incert_invalides == 0:
    print("OK : aucune valeur physiquement impossible.")
else:
    display(hors_plage)
```

<h3 id="s10-2">10.2. Détection statistique par pays et par mois<span class="up"><a href="#section-10">↑ section 10</a><a href="#sommaire">☰ sommaire</a></span></h3>
<div class="card"><p>Deuxième niveau : des valeurs statistiquement atypiques. Comparer toutes les températures entre elles serait trompeur, car elles varient avec les saisons. On applique donc la règle de l'IQR séparément pour chaque pays et chaque mois calendaire (par exemple tous les mois de janvier du Bénin).</p></div>
</div>

```python
df_net["mois"] = df_net["dt"].dt.month

# Quartiles de chaque groupe pays × mois, répétés sur chaque ligne du groupe
g = df_net.groupby(["Country", "mois"])["AverageTemperature"]
q1 = g.transform(lambda s: s.quantile(0.25))
q3 = g.transform(lambda s: s.quantile(0.75))
mediane = g.transform("median")
iqr = q3 - q1

# True si la température sort de [Q1 - 1,5 IQR ; Q3 + 1,5 IQR] de son groupe (les NaN donnent False)
t = df_net["AverageTemperature"]
df_net["temp_atypique"] = (t < q1 - 1.5 * iqr) | (t > q3 + 1.5 * iqr)
df_net["ecart_mediane"] = (t - mediane).round(3)

print("Températures atypiques :", df_net["temp_atypique"].sum(),
      f"({df_net['temp_atypique'].mean() * 100:.2f} % des lignes)")
display(df_net.groupby("Country")["temp_atypique"].sum().rename("atypiques").to_frame())

# Les 10 cas les plus éloignés de la médiane de leur pays et de leur mois
cas = df_net[df_net["temp_atypique"]].copy()
cas["ecart_abs"] = cas["ecart_mediane"].abs()
display(cas.sort_values("ecart_abs", ascending=False)
        [["Country", "dt", "AverageTemperature", "AverageTemperatureUncertainty", "ecart_mediane"]].head(10))
```

```python
# Visualisation : boîtes à moustaches par mois, une figure par pays
pays_liste = sorted(df_net["Country"].unique())
fig, axes = plt.subplots(2, 3, figsize=(16, 8), sharey=True)
for ax, pays in zip(axes.ravel(), pays_liste):
    sns.boxplot(data=df_net[df_net["Country"] == pays], x="mois", y="AverageTemperature", color="#9fd3dc", ax=ax)
    ax.set_title(pays)
    ax.set_xlabel("Mois")
    ax.set_ylabel("Température (°C)")
for ax in axes.ravel()[len(pays_liste):]:
    ax.axis("off")                           # masque la case vide de la grille 2×3
plt.suptitle("Températures mensuelles par pays (valeurs atypiques = points isolés)", y=1.02)
plt.tight_layout()
plt.show()

# Profil des températures atypiques : chaudes ou froides ? anciennes ou récentes ? incertaines ?
atypiques = df_net[df_net["temp_atypique"]]
print("Atypiques chaudes (au-dessus de la médiane) :", int((atypiques["ecart_mediane"] > 0).sum()),
      "| froides :", int((atypiques["ecart_mediane"] < 0).sum()))
print("Atypiques avant 1950 :", int((atypiques["annee"] < 1950).sum()),
      "| depuis 1950 :", int((atypiques["annee"] >= 1950).sum()))
print("Atypiques ayant aussi une forte incertitude :", int(atypiques["incertitude_elevee"].sum()))
```

<div class="success"><strong>Décision :</strong> aucune valeur n'est supprimée ni modifiée. Les températures atypiques sont des signaux statistiques, pas des erreurs prouvées : un mois exceptionnellement chaud peut être réel, et le réchauffement des dernières décennies peut faire sortir des valeurs récentes de la fourchette historique. <h3 id="s11-2">11.2. Tableau comparatif<span class="up"><a href="#section-11">↑ section 11</a><a href="#sommaire">☰ sommaire</a></span></h3>
<div class="card"><p>Les mêmes indicateurs qu'au diagnostic initial sont recalculés sur le jeu final : l'effet du nettoyage se lit colonne par colonne.</p></div>

```python
def dates_invalides(d):
    """Nombre de dates illisibles, que la colonne soit en texte ou en datetime."""
    return int(pd.to_datetime(d["dt"], errors="coerce").isna().sum())

lignes = [
    ("Nombre de lignes",                         len(df_etude), len(df_final)),
    ("Nombre de colonnes",                       df_etude.shape[1], df_final.shape[1]),
    ("Type de la colonne dt",                    str(df_etude["dt"].dtype), str(df_final["dt"].dtype)),
    ("Dates invalides",                          dates_invalides(df_etude), dates_invalides(df_final)),
    ("Libellés de pays distincts",               df_etude["Country"].nunique(), df_final["Country"].nunique()),
    ("Libellés utilisés",                        ", ".join(sorted(df_etude["Country"].unique())),
                                                 ", ".join(sorted(df_final["Country"].unique()))),
    ("Températures manquantes",                  int(df_etude["AverageTemperature"].isna().sum()), int(df_final["AverageTemperature"].isna().sum())),
    ("Taux de températures manquantes (%)",     round(df_etude["AverageTemperature"].isna().mean() * 100, 2), round(df_final["AverageTemperature"].isna().mean() * 100, 2)),
    ("Incertitudes manquantes",                  int(df_etude["AverageTemperatureUncertainty"].isna().sum()), int(df_final["AverageTemperatureUncertainty"].isna().sum())),
    ("Doublons exacts (4 colonnes d'origine)",  int(df_etude[colonnes_origine].duplicated().sum()), int(df_final[colonnes_origine].duplicated().sum())),
    ("Doublons pays-date",                       int(df_etude.duplicated(subset=["Country", "dt"]).sum()), int(df_final.duplicated(subset=["Country", "dt"]).sum())),
    ("Première date",                            str(df_etude["dt"].min()), str(df_final["dt"].min().date())),
    ("Dernière date",                            str(df_etude["dt"].max()), str(df_final["dt"].max().date())),
    ("Température minimale (°C)",                round(df_etude["AverageTemperature"].min(), 3), round(df_final["AverageTemperature"].min(), 3)),
    ("Température maximale (°C)",                round(df_etude["AverageTemperature"].max(), 3), round(df_final["AverageTemperature"].max(), 3)),
]
# Valeurs douteuses : même fonction (définie en 5.5) appliquée avant et après
anomalies_finales = compter_anomalies(df_final)
for nom in anomalies_initiales:
    lignes.append((nom, anomalies_initiales[nom], anomalies_finales[nom]))
comparatif = pd.DataFrame(lignes, columns=["Indicateur", "Avant nettoyage", "Après nettoyage"]).astype(str)
display(comparatif)
```

```python
pivot_manq = (df_final.assign(decennie=df_final["annee"] // 10 * 10,
                               manquante=df_final["temperature_manquante"].astype(float))
              .pivot_table(index="Country", columns="decennie", values="manquante", aggfunc="mean") * 100)

plt.figure(figsize=(14, 4.5))
sns.heatmap(pivot_manq, annot=True, fmt=".0f", cmap="YlOrRd", vmin=0, vmax=100,
            cbar_kws={"label": "Températures manquantes (%)"})
plt.title("Mois sans mesure par pays et par décennie")
plt.xlabel("Décennie")
plt.ylabel("")
plt.tight_layout()
plt.show()
```

<h3 id="s12-2">12.2. Températures annuelles moyennes<span class="up"><a href="#section-12">↑ section 12</a><a href="#sommaire">☰ sommaire</a></span></h3>
<div class="card"><p>Moyenne annuelle des températures, calculée uniquement sur les années ayant 12 mois de mesure (une moyenne annuelle sur des mois manquants serait biaisée par la saisonnalité). La courbe épaisse est une moyenne glissante sur 10 ans.</p></div>
</div>

```python
annuel = (df_final.groupby(["Country", "annee"])
          .agg(mois_valides=("AverageTemperature", "count"),
               temperature_moyenne=("AverageTemperature", "mean"))
          .reset_index())
# Une année est « complète » si elle a 12 mois de mesure ; sinon on met NaN (pas de moyenne biaisée)
annuel.loc[annuel["mois_valides"] < 12, "temperature_moyenne"] = np.nan

# Tableau année × pays sur TOUTES les années : les années sans valeur restent NaN,
# donc les courbes sont interrompues au lieu d'être reliées par un trait trompeur
serie = (annuel.pivot(index="annee", columns="Country", values="temperature_moyenne")
         .reindex(range(df_final["annee"].min(), df_final["annee"].max() + 1)))
lissee = serie.rolling(10, min_periods=7).mean()          # moyenne glissante sur 10 années calendaires

palette = dict(zip(sorted(df_final["Country"].unique()), sns.color_palette("tab10", 5)))

fig, ax = plt.subplots(figsize=(13, 6))
for pays in serie.columns:
    ax.plot(serie.index, serie[pays], color=palette[pays], alpha=0.3, linewidth=1)
    ax.plot(lissee.index, lissee[pays], color=palette[pays], linewidth=2.5, label=pays)
ax.set_title("Température annuelle moyenne (années complètes) et moyenne glissante sur 10 ans")
ax.set_xlabel("Année")
ax.set_ylabel("Température (°C)")
ax.legend(title="Pays", bbox_to_anchor=(1.02, 1), loc="upper left")
plt.tight_layout()
plt.show()

# Écart entre la période 1900-1929 et la période 1983-2012 (années complètes, donc sans manquants)
ancien = serie.loc[1900:1929].mean()
recent = serie.loc[1983:2012].mean()
display(pd.DataFrame({"1900-1929 (°C)": ancien, "1983-2012 (°C)": recent, "hausse (°C)": recent - ancien}).round(2))
```

<h3 id="s13-2">13.2. Relecture et contrôle<span class="up"><a href="#section-13">↑ section 13</a><a href="#sommaire">☰ sommaire</a></span></h3>
<div class="card"><p>Un export n'est valide que s'il se relit à l'identique : on recharge le CSV et on le compare au jeu final (dimensions, dates, manquants et colonnes).</p></div>
</div>

```python
relu = pd.read_csv(fichier_sortie, parse_dates=["dt"])

controles = {
    "Mêmes dimensions": relu.shape == df_final.shape,
    "Mêmes colonnes": list(relu.columns) == list(df_final.columns),
    "Dates bien relues (datetime)": pd.api.types.is_datetime64_any_dtype(relu["dt"]),
    "Mêmes températures manquantes": relu["AverageTemperature"].isna().sum() == df_final["AverageTemperature"].isna().sum(),
    "Cinq pays exactement": relu["Country"].nunique() == 5,
    "Accents conservés": {"Bénin", "Côte d'Ivoire", "Sénégal"}.issubset(set(relu["Country"])),
}
for nom, ok in controles.items():
    print("OK         " if ok else "A VERIFIER ", nom)

relu.head()
```

<h3 id="s14-2">14.2. Conclusion<span class="up"><a href="#section-14">↑ section 14</a><a href="#sommaire">☰ sommaire</a></span></h3>
<div class="objective"><p>Nous sommes partis du fichier brut <code>GlobalLandTemperaturesByCountry.csv</code> et nous livrons un jeu propre et documenté pour cinq pays d'Afrique de l'Ouest : Bénin, Burkina Faso, Côte d'Ivoire, Ghana et Sénégal. Les dates ont été converties et vérifiées, les noms harmonisés et les doublons recherchés (il n'y en avait pas).</p>
<p>Surtout, nous n'avons supprimé ni inventé aucune mesure. Les 775 mois sans mesure (7,84 % des lignes, presque tous avant 1900) restent vides mais sont signalés. Les 212 mois à forte incertitude (2,14 %, tous avant 1930) et les 81 températures atypiques (0,82 %) sont simplement marqués : chacun peut les garder, les pondérer ou les écarter selon son analyse.</p>
<p>En résumé, les cinq séries sont complètes et fiables de 1900 à août 2013, l'incertitude passe de plus de 1 °C au XIXᵉ siècle à environ 0,25 °C vers 1950, et les moyennes annuelles montent nettement depuis les années 1980. C'est la réponse à notre question de départ (<a href="#s01-2">1.2</a>) : mieux vaut signaler que corriger à l'aveugle.</p></div>
</div>

<h3 id="s14-3">14.3. Limites et perspectives<span class="up"><a href="#section-14">↑ section 14</a><a href="#sommaire">☰ sommaire</a></span></h3>
<div class="card"><ul>
<li>Manquants anciens : avant 1900, jusqu'à 79 % des mois manquent dans certaines décennies (Bénin, Ghana) ; pour une analyse de tendance, privilégier la période 1900-2012 (années complètes, voir <a href="#s12-2">12.2</a>). L'année 2013 est incomplète (septembre sans mesure).</li>
<li>Seuils choisis : la plage physique (10–45 °C) et la règle de Tukey (1,5 × IQR) sont des choix raisonnables mais discutables ; ils sont modifiables en tête de leurs cellules.</li>
<li>Incertitudes : les mesures anciennes, plus incertaines, pourraient être pondérées dans un modèle statistique.</li>
<li>Perspectives : comparer les tendances entre pays, étudier l'évolution du réchauffement par décennie ou croiser avec d'autres sources climatiques.</li>
</ul></div>
